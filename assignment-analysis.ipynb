```python
#Data analysis python assignment
#import liberaries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#Make plots look nicer
sns.set_palette ("Set2")

#=====================task 1:Load and explore the dataset========================= 
from sklearn.datasets import load_iris

#in my work i have used classic iris dataset
iris_data = load_iris(as_frame = True)
df = iris_data.frame

#preview first rows
print("First 5 rows of the dataset are:")
display(df.head())

#check structure and missing values
print("\n Dataset info:")
print(df.info())

print("\n Missing values per column:")
print(df.isnull().sum())

#In Iris there are no missing values, but if there were i could:
#df = df.dropna()

#==========================Task 2:basic Data analysis=====================================
print("\n Statistical Summary: ")
display(df.describe())

#Grouping: Mean petal length per species
grouped = df.groupby("target")["petal length (cm)"].mean()
print("\n Average petal Length per species: ")
display(grouped) 

#Add species name for readability
df["species"] = df["target"].map({i: name for i, name in enumerate(iris_data.target_names)})

#patterns/insights

print("\n Insigts:")
print("I have seen that Iris-virginica generally has the largest patal length,While Iris-setosa has the smallest. This difference is clear in stats")


#========================Task 3: Data vissualization============================
#---------1. Line chart: sepal length trend(just taking index as "time")


plt.figure(figsize = (8,5))
plt.plot(df.index,df["sepal length (cm)"],
        label="Sapal Length")
plt.title("Line Chart : sapel Length over Sample")
plt.xlabel("Sample Index")
plt.ylabel("Sapel Length (cm)")
plt.show()

#2----------.Bar chart:average petal length per species

plt.figure(figsize=(8,5))
grouped.plot(kind = "bar" , color = ["#ff9999","#66b3ff","#99ff99"])
plt.title("Bar Chart: Average Petal Length per Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")
plt.show()

#-------------3. Histogram : distribution of sepal width
plt.figure(figsize = (8,5))
plt.hist(df["sepal width (cm)"],bins = 20,color ="skyblue",edgecolor="black" )
plt.title("Histogram: Sepal width Distribution")
plt.xlabel("Sepal width (cm)")
plt.ylabel("Frequency")
plt.show()

#--------------4. Scatter plot: sepal length vs petal
plt.figure(figsize = (8,5))
sns.scatterplot(x="sepal length (cm)",y="petal length (cm)",hue= "species",data=df)
plt.title("Scatter Plot: Sepal Length vs Petal")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.show()
```
